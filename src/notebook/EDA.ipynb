```python
import pandas as pd

df = pd.read_csv('EDA.csv')

# Features aur Target alag karo
final_features = ['age','diameter_cm','gender_M','smoker','pack_year','hypertension','diabetes']
X = df[final_features]
y = df['ruptured']

# Pehle NaN check karo
print("Y me null values:", y.isnull().sum())

# Step 1: NaN fill karo, phir int banao
y = y.fillna(0).astype(int)   # NaN ko 0 = Not Ruptured maan liya
X = X.fillna(X.mean())        # X ke NaN ko mean se fill kiya

print("X shape:", X.shape)
print("y distribution:\n", y.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(class_weight='balanced', max_iter=1000, solver='liblinear')
model.fit(X, y)

# Feature importance
importance = pd.DataFrame({'Feature': X.columns, 'Coef': model.coef_[0]})
importance['Abs_Coef'] = importance['Coef'].abs()
importance = importance.sort_values('Abs_Coef', ascending=False)
print(importance[['Feature','Coef']])

# Predict kar ke dekh
print("Train Accuracy:", model.score(X, y))
print("Predictions:", model.predict(X))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
scores = cross_val_score(model, X, y, cv=cv, scoring='roc_auc')
print("CV ROC-AUC:", scores.mean())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

model = LogisticRegression(class_weight='balanced', max_iter=1000, solver='liblinear')
model.fit(X, y)

print("Train accuracy:", model.score(X, y))
print("\nConfusion Matrix:\n", confusion_matrix(y, model.predict(X)))
print("\nReport:\n", classification_report(y, model.predict(X)))

# Feature importance
importance = pd.DataFrame({'Feature': X.columns, 'Coef': model.coef_[0]})
print(importance.sort_values('Coef', key=abs, ascending=False))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=2, shuffle=True, random_state=42)
scores = cross_val_score(model, X, y, cv=cv, scoring='roc_auc')
print("ROC-AUC:", scores.mean())
```
